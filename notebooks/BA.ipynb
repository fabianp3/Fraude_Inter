**Variables que quedan para el modelo**

```python
from pathlib import Path
import sys

root_dir = Path.cwd().resolve().parent if Path.cwd().name == 'notebooks' else Path.cwd().resolve()
if str(root_dir) not in sys.path:
    sys.path.append(str(root_dir))

from funciones.variables_modelo import (
    VARIABLES_CATEGORICAS_MODELO,
    VARIABLES_NUMERICAS,
    VARIABLES_TOTALES_MODELO,
 )

variables_numericas = VARIABLES_NUMERICAS.copy()
variables_categoricas = VARIABLES_CATEGORICAS_MODELO.copy()
variables_totales = VARIABLES_TOTALES_MODELO.copy()
```

Para seleccionar las variables con mayor poder discriminativo se combinan tres criterios: Cramér's V para variables categóricas, correlación de Pearson (en valor absoluto) para variables numéricas, y Mutual Information, que aplica por igual a ambos tipos de variable frente a `FRAUDE`. Cada variable se ordena según su métrica nativa y según su Mutual Information; el promedio de ambos rankings define un score combinado que permite comparar variables numéricas y categóricas en una misma escala.

```python
from scipy.stats import chi2_contingency

def cramers_v(x, y):
    tabla = pd.crosstab(x, y)
    chi2 = chi2_contingency(tabla)[0]
    n = tabla.sum().sum()
    phi2 = chi2 / n
    r, k = tabla.shape
    phi2corr = max(0, phi2 - ((k - 1) * (r - 1)) / (n - 1))
    rcorr = r - ((r - 1) ** 2) / (n - 1)
    kcorr = k - ((k - 1) ** 2) / (n - 1)
    return np.sqrt(phi2corr / min((kcorr - 1), (rcorr - 1)))

cramers_scores = {col: cramers_v(df1[col], df1['FRAUDE']) for col in variables_categoricas}
cramers_scores = pd.Series(cramers_scores, name='Cramers_V').sort_values(ascending=False)
cramers_scores
```

```python
correlacion_scores = (
    df1[variables_numericas + ['FRAUDE']]
    .corr()['FRAUDE']
    .drop('FRAUDE')
    .abs()
    .sort_values(ascending=False)
)
correlacion_scores.name = 'Correlacion_Abs'
correlacion_scores
```

```python
from sklearn.feature_selection import mutual_info_classif
from sklearn.preprocessing import LabelEncoder

X_mi = df1[variables_totales].copy()
for col in variables_categoricas:
    X_mi[col] = LabelEncoder().fit_transform(X_mi[col].astype(str))

es_discreta = [col in variables_categoricas for col in variables_totales]

mi_scores = mutual_info_classif(
    X_mi, df1['FRAUDE'], discrete_features=es_discreta, random_state=42
)
mi_scores = pd.Series(mi_scores, index=variables_totales, name='Mutual_Info').sort_values(ascending=False)
mi_scores
```

```python
tabla_discriminativa = pd.DataFrame(index=variables_totales)
tabla_discriminativa['Tipo'] = ['Numerica' if v in variables_numericas else 'Categorica' for v in variables_totales]
tabla_discriminativa['Cramers_V'] = cramers_scores
tabla_discriminativa['Correlacion_Abs'] = correlacion_scores
tabla_discriminativa['Mutual_Info'] = mi_scores

tabla_discriminativa['Metrica_Nativa'] = tabla_discriminativa['Cramers_V'].fillna(tabla_discriminativa['Correlacion_Abs'])
tabla_discriminativa['Rank_MI'] = tabla_discriminativa['Mutual_Info'].rank(ascending=False)
tabla_discriminativa['Rank_Nativo'] = tabla_discriminativa['Metrica_Nativa'].rank(ascending=False)
tabla_discriminativa['Score_Combinado'] = tabla_discriminativa[['Rank_MI', 'Rank_Nativo']].mean(axis=1)

tabla_discriminativa = tabla_discriminativa.sort_values('Score_Combinado')
tabla_discriminativa
```

```python
N_VARIABLES = 10
variables_seleccionadas = tabla_discriminativa.head(N_VARIABLES).index.tolist()

df_variables_discriminativas = df1[variables_seleccionadas + ['FRAUDE']].copy()
df_variables_discriminativas.to_csv('../datos/procesados/dataset_variables_discriminativas.csv', index=False)

print(f"Variables seleccionadas ({N_VARIABLES}): {variables_seleccionadas}")
df_variables_discriminativas.head()
```

El score combinado (promedio del ranking por Mutual Information y del ranking por la métrica nativa de cada tipo de variable) coincide con lo observado en el análisis exploratorio: las variables más discriminativas son señales de comportamiento reciente y de seguridad (`DISPOSITIVO_NUEVO`, `BENEFICIARIO_NUEVO`, `CAMBIO_CLAVE_RECIENTE`), el número de intentos fallidos (`NUM_INTENTOS_FALLIDOS`), y el contexto de la operación (`CANAL`, `OPERACION_NOCTURNA`, `PAIS_DESTINO`, `TIPO_TRANSACCION`), junto con `MONTO_TRANSACCION` y `DISTANCIA_UBICACION_KM` como variables numéricas relevantes. Variables demográficas como `SEXO`, `TIPO_CLIENTE` y `EDAD` quedan en las últimas posiciones, confirmando su bajo poder predictivo.
